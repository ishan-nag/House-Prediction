Feature Engineering

```python
# Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Extract temporal features
df['Year_Sold'] = df['Date'].dt.year
df['Month_Sold'] = df['Date'].dt.month
df['Quarter_Sold'] = df['Date'].dt.quarter
df['Week_of_Year'] = df['Date'].dt.isocalendar().week
df['Day_of_Week'] = df['Date'].dt.dayofweek  # Monday=0, Sunday=6
df['Day_of_Month'] = df['Date'].dt.day

# Create binary features for Indian seasons/festive periods
# Indian festive months (Diwali, Dussehra, etc.) - typically Sep-Nov
df['Festive_Month'] = df['Month_Sold'].isin([9, 10, 11]).astype(int)

# Financial quarters (India: Apr-Mar)
df['Financial_Quarter'] = ((df['Month_Sold'] - 4) % 12 // 3 + 1).where(df['Month_Sold'] >= 4, 
                                                                       ((df['Month_Sold'] + 8) % 12 // 3 + 1))

# Pre-monsoon (Mar-May) and Post-monsoon (Oct-Dec) seasons
df['Pre_Monsoon'] = df['Month_Sold'].isin([3, 4, 5]).astype(int)
df['Post_Monsoon'] = df['Month_Sold'].isin([10, 11, 12]).astype(int)

# Year-Month combination for trend analysis
df['YearMonth'] = df['Year_Sold'] * 100 + df['Month_Sold']

# Days since reference date (for time series trend)
df['Days_Since_Start'] = (df['Date'] - df['Date'].min()).dt.days

# Month sine/cosine for cyclical encoding
df['Month_Sin'] = np.sin(2 * np.pi * df['Month_Sold'] / 12)
df['Month_Cos'] = np.cos(2 * np.pi * df['Month_Sold'] / 12)

# Create season feature
def get_indian_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Summer'
    elif month in [6, 7, 8, 9]:
        return 'Monsoon'
    else:
        return 'Post-Monsoon'

df['Season'] = df['Month_Sold'].apply(get_indian_season)

# One-hot encode season
df = pd.get_dummies(df, columns=['Season'], prefix='Season')

# Check temporal patterns
print(df[['Date', 'Year_Sold', 'Month_Sold', 'Quarter_Sold', 'Festive_Month']].head(10))

# Optional: Analyze price trends over time
monthly_avg_price = df.groupby('YearMonth')['Price'].mean()
print("\nMonthly Average Price Trend:")
print(monthly_avg_price.tail(12))
```

```python
# Get current year
current_year = pd.Timestamp.now().year

# House Age
df['House_Age'] = current_year - df['Built Year']

# Years Since Renovation (handle 0 or NaN for no renovation)
df['Years_Since_Renov'] = current_year - df['Renovation Year']
# Replace negative or unrealistic values with 0 (no renovation)
df['Years_Since_Renov'] = df['Years_Since_Renov'].clip(lower=0)

# Binary renovation flag
df['Is_Renovated'] = (df['Renovation Year'] > df['Built Year']).astype(int)

# Age bins for categorical encoding
df['Age_Bin'] = pd.cut(df['House_Age'], 
                       bins=[0, 5, 10, 20, 30, 50, 100, 200],
                       labels=['0-5', '5-10', '10-20', '20-30', '30-50', '50-100', '100+'])

# Interaction between age and renovation
df['Age_Renovated_Ratio'] = df['Years_Since_Renov'] / df['House_Age'].replace(0, 1)  # Avoid division by zero

# Recently renovated flag (within last 5 years)
df['Recently_Renovated'] = (df['Years_Since_Renov'] <= 5).astype(int)

# Old but renovated flag
df['Old_Renovated'] = ((df['House_Age'] > 30) & (df['Is_Renovated'] == 1)).astype(int)

# Never renovated old house
df['Old_No_Renov'] = ((df['House_Age'] > 30) & (df['Is_Renovated'] == 0)).astype(int)

# Check for anomalies
print(f"Houses with negative age: {(df['House_Age'] < 0).sum()}")
print(f"Houses older than 200 years: {(df['House_Age'] > 200).sum()}")
print(f"Renovated houses: {df['Is_Renovated'].sum()} ({df['Is_Renovated'].mean()*100:.1f}%)")

# Display sample
print("\nSample of age features:")
print(df[['Built Year', 'Renovation Year', 'House_Age', 'Years_Since_Renov', 'Is_Renovated']].head(10))
```

```python
# Price per SqFt
df['Price_per_SqFt'] = df['Price'] / df['living area']

# Handle infinite or extremely high values (set to reasonable limits)
df['Price_per_SqFt'] = df['Price_per_SqFt'].replace([np.inf, -np.inf], np.nan)
upper_limit = df['Price_per_SqFt'].quantile(0.99)
df['Price_per_SqFt'] = df['Price_per_SqFt'].clip(upper=upper_limit)

# Built-up Ratio (plot utilization)
df['Built_up_Ratio'] = df['living area'] / df['lot area']

# Handle division by zero and extreme values
df['Built_up_Ratio'] = df['Built_up_Ratio'].replace([np.inf, -np.inf], np.nan)
df['Built_up_Ratio'] = df['Built_up_Ratio'].clip(upper=1.0)  # Can't exceed 1.0 for plot utilization

# Bed-Bath Ratio
df['Bed_Bath_Ratio'] = df['number of bathrooms'] / df['number of bedrooms']

# Handle division by zero (if bedrooms = 0, set ratio to bathrooms count)
df['Bed_Bath_Ratio'] = df.apply(
    lambda row: row['number of bathrooms'] if row['number of bedrooms'] == 0 else row['Bed_Bath_Ratio'],
    axis=1
)

# Total Area
df['Total_Area'] = df['Area of the house(excluding basement)'] + df['Area of the basement']

# Additional space efficiency features
# Floor area ratio (if you have number of floors)
df['Area_per_Floor'] = df['living area'] / df['number of floors'].replace(0, 1)

# Compactness indicator (small lot with good living area)
df['Compact_House'] = ((df['lot area'] < df['lot area'].median()) & 
                       (df['living area'] > df['living area'].median())).astype(int)

# Luxury indicator (large bathrooms relative to bedrooms)
df['Luxury_Bath'] = (df['Bed_Bath_Ratio'] > 1.5).astype(int)

# Space utilization efficiency
df['Space_Efficiency'] = df['living area'] / df['Total_Area']

# Room density
df['Rooms_per_SqFt'] = (df['number of bedrooms'] + df['number of bathrooms']) / df['living area']

# Plot coverage percentage
df['Plot_Coverage_Pct'] = (df['living area'] / df['lot area']) * 100
df['Plot_Coverage_Pct'] = df['Plot_Coverage_Pct'].clip(upper=100)

# Display summary
print("Space & Layout Features Summary:")
print("="*50)
print(f"Average Price per SqFt: ₹{df['Price_per_SqFt'].mean():.0f}")
print(f"Average Built-up Ratio: {df['Built_up_Ratio'].mean():.2f}")
print(f"Average Bed-Bath Ratio: {df['Bed_Bath_Ratio'].mean():.2f}")
print(f"Average Total Area: {df['Total_Area'].mean():.0f} sq.ft")
print(f"Compact Houses: {df['Compact_House'].sum()} ({df['Compact_House'].mean()*100:.1f}%)")
print(f"Luxury Bath Houses: {df['Luxury_Bath'].sum()} ({df['Luxury_Bath'].mean()*100:.1f}%)")

# Check for anomalies
print("\nAnomaly Checks:")
print(f"Houses with 0 bedrooms: {(df['number of bedrooms'] == 0).sum()}")
print(f"Houses with Built-up Ratio > 1: {(df['Built_up_Ratio'] > 1).sum()}")
print(f"Houses with 0 living area: {(df['living area'] == 0).sum()}")
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Option 1: Neighborhood Clustering using K-Means
# Prepare coordinates for clustering
coords = df[['Lattitude', 'Longitude']].copy()

# Scale coordinates for better clustering
scaler = StandardScaler()
coords_scaled = scaler.fit_transform(coords)

# Determine optimal number of clusters using elbow method
wcss = []
cluster_range = range(2, 15)
for n_clusters in cluster_range:
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    kmeans.fit(coords_scaled)
    wcss.append(kmeans.inertia_)

# Plot elbow curve to choose optimal clusters
plt.figure(figsize=(10, 6))
plt.plot(cluster_range, wcss, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.title('Elbow Method for Optimal Clusters')
plt.grid(True)
plt.show()

# Apply K-Means with chosen number of clusters (e.g., 8)
optimal_clusters = 8  # Adjust based on elbow plot
kmeans = KMeans(n_clusters=optimal_clusters, random_state=42, n_init=10)
df['Neighborhood_Cluster'] = kmeans.fit_predict(coords_scaled)

# Option 2: Area Average Price per Postal Code
# Calculate mean price per postal code from TRAINING data only
# We'll do this properly later during train-test split, but for now:
postal_avg_price = df.groupby('Postal Code')['Price'].transform('mean')
df['Area_Avg_Price'] = postal_avg_price

# Option 3: Combine both approaches
# Calculate average price within each cluster
cluster_avg_price = df.groupby('Neighborhood_Cluster')['Price'].transform('mean')
df['Cluster_Avg_Price'] = cluster_avg_price
```

```python
# Luxury & Premium Indicators

# High Grade indicator (assuming grade is on scale 1-10)
grade_threshold = 8
df['High_Grade'] = (df['grade of the house'] > grade_threshold).astype(int)

# Good Condition indicator (assuming condition is on scale 1-5)
condition_threshold = 4
df['Good_Condition'] = (df['condition of the house'] > condition_threshold).astype(int)

# Premium Grade-Condition combination
df['Premium_Property'] = (df['High_Grade'] == 1) & (df['Good_Condition'] == 1)
df['Premium_Property'] = df['Premium_Property'].astype(int)

# Grade categories
df['Grade_Category'] = pd.cut(df['grade of the house'], 
                              bins=[0, 5, 7, 10], 
                              labels=['Low', 'Medium', 'High'])

# Amenity & Accessibility Features

# Schools density proxy (using lot area as proxy for density)
df['Schools_per_SqKm'] = df['Number of schools nearby'] / (df['lot area'] + 1) * 10000
# Handle extreme values
df['Schools_per_SqKm'] = df['Schools_per_SqKm'].clip(upper=df['Schools_per_SqKm'].quantile(0.99))

# Airport accessibility (inverse relationship)
df['Airport_Accessibility'] = 1 / (df['Distance from the airport'] + 0.1)  # Add small constant to avoid division by zero

# Distance categories
df['Airport_Distance_Category'] = pd.cut(df['Distance from the airport'],
                                        bins=[0, 10, 25, 50, 100, float('inf')],
                                        labels=['Very Near', 'Near', 'Moderate', 'Far', 'Very Far'])

# Schools nearby categories
df['Schools_Category'] = pd.cut(df['Number of schools nearby'],
                               bins=[-1, 0, 3, 7, 15, float('inf')],
                               labels=['None', 'Few', 'Moderate', 'Many', 'Excellent'])

# Good accessibility indicator (both airport and schools)
df['Good_Accessibility'] = ((df['Distance from the airport'] < 30) & 
                           (df['Number of schools nearby'] > 5)).astype(int)

# Check created features
print("Luxury & Amenity Features Summary:")
print("="*50)
print(f"High Grade houses (>8): {df['High_Grade'].sum()} ({df['High_Grade'].mean()*100:.1f}%)")
print(f"Good Condition houses (>4): {df['Good_Condition'].sum()} ({df['Good_Condition'].mean()*100:.1f}%)")
print(f"Premium Properties (High Grade & Good Condition): {df['Premium_Property'].sum()} ({df['Premium_Property'].mean()*100:.1f}%)")
print(f"Average schools per sq.km proxy: {df['Schools_per_SqKm'].mean():.2f}")
print(f"Average airport accessibility: {df['Airport_Accessibility'].mean():.4f}")
print(f"Good Accessibility houses: {df['Good_Accessibility'].sum()} ({df['Good_Accessibility'].mean()*100:.1f}%)")

# Display distribution
print("\nGrade Categories:")
print(df['Grade_Category'].value_counts(normalize=True).round(3))
print("\nAirport Distance Categories:")
print(df['Airport_Distance_Category'].value_counts(normalize=True).round(3))
print("\nSchools Categories:")
print(df['Schools_Category'].value_counts(normalize=True).round(3))

# Quick check for anomalies
print(f"\nAnomaly Checks:")
print(f"Houses with 0 schools nearby: {(df['Number of schools nearby'] == 0).sum()}")
print(f"Houses with 0 distance to airport: {(df['Distance from the airport'] == 0).sum()}")
print(f"Max schools nearby: {df['Number of schools nearby'].max()}")
print(f"Max airport distance: {df['Distance from the airport'].max()}")
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

df_copy = df.copy()

original_price = df_copy['Price'].copy()

df_copy['Price_log'] = np.log1p(df_copy['Price'])

skewed_columns = ['living area', 'lot area', 'Area of the house(excluding basement)', 
                  'Area of the basement', 'Total_Area', 'Price_per_SqFt',
                  'Number of schools nearby', 'Schools_per_SqKm']

for col in skewed_columns:
    if col in df_copy.columns:
        if df_copy[col].min() >= 0:
            df_copy[f'{col}_log'] = np.log1p(df_copy[col])
        else:
            df_copy[f'{col}_log'] = np.log(df_copy[col] - df_copy[col].min() + 1)

numerical_features = df_copy.select_dtypes(include=[np.number]).columns.tolist()
target_cols = ['Price', 'Price_log', 'Price_per_SqFt', 'Price_per_SqFt_log']
numerical_features = [col for col in numerical_features if col not in target_cols]

scaler = StandardScaler()
scaled_array = scaler.fit_transform(df_copy[numerical_features])
df_scaled = pd.DataFrame(scaled_array, columns=[f'{col}_scaled' for col in numerical_features], index=df_copy.index)
df_copy = pd.concat([df_copy, df_scaled], axis=1)

categorical_cols = ['Grade_Category', 'Age_Bin', 'Airport_Distance_Category', 
                    'Schools_Category', 'Season_Winter', 'Season_Summer', 
                    'Season_Monsoon', 'Season_Post-Monsoon']
categorical_cols = [col for col in categorical_cols if col in df_copy.columns]

for col in categorical_cols:
    if col in df_copy.columns:
        dummies = pd.get_dummies(df_copy[col], prefix=col, drop_first=True)
        df_copy = pd.concat([df_copy, dummies], axis=1)

features_to_drop = ['id', 'Date', 'Price', 'Price_per_SqFt', 'Price_per_SqFt_log']
for col in skewed_columns:
    features_to_drop.extend([col, f'{col}_log'])

features_to_drop = [col for col in features_to_drop if col in df_copy.columns]
df_copy = df_copy.drop(columns=features_to_drop)

binary_cols = ['waterfront present', 'Is_Premium_Locality', 'Is_Statistically_Premium',
               'Is_Premium_Cluster', 'Premium_Property', 'Festive_Month',
               'Pre_Monsoon', 'Post_Monsoon', 'Is_Renovated', 'Recently_Renovated',
               'Old_Renovated', 'Old_No_Renov', 'Compact_House', 'Luxury_Bath',
               'High_Grade', 'Good_Condition', 'Good_Accessibility']

for col in binary_cols:
    if col in df_copy.columns:
        df_copy[col] = df_copy[col].astype(int)

final_features = [col for col in df_copy.columns if col != 'Price_log']
X = df_copy[final_features]
y = df_copy['Price_log']

print(f"Final dataset shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"Number of features: {len(X.columns)}")

selected_features = ['living area_scaled', 'lot area_scaled', 'number of bedrooms',
                     'number of bathrooms', 'Total_Area_scaled', 'Price_per_SqFt_scaled',
                     'House_Age', 'Years_Since_Renov', 'Built_up_Ratio',
                     'Bed_Bath_Ratio', 'Area_Avg_Price_scaled', 'Cluster_Avg_Price_scaled',
                     'Neighborhood_Cluster', 'High_Grade', 'Good_Condition',
                     'Schools_per_SqKm_scaled', 'Airport_Accessibility_scaled']

selected_features = [col for col in selected_features if col in X.columns]
print(f"Selected {len(selected_features)} key features for XGBoost")

X_final = X[selected_features]
```

```python
from sklearn.model_selection import train_test_split

X_temp = X_final.copy()
y_temp = y.copy()

# Manual target encoding for Postal Code without using category_encoders
if 'Postal Code' in df.columns:
    X_temp['Postal_Code'] = df['Postal Code'].iloc[X_temp.index]
    
    X_train_temp, X_test_temp, y_train_temp, y_test_temp = train_test_split(
        X_temp, y_temp, test_size=0.2, random_state=42, shuffle=True
    )
    
    # Manual target encoding
    postal_target_map = y_train_temp.groupby(X_train_temp['Postal_Code']).mean()
    
    X_train_temp['Postal_Code_Encoded'] = X_train_temp['Postal_Code'].map(postal_target_map)
    X_test_temp['Postal_Code_Encoded'] = X_test_temp['Postal_Code'].map(postal_target_map)
    
    # Handle unmapped postal codes in test set
    train_mean = y_train_temp.mean()
    X_train_temp['Postal_Code_Encoded'] = X_train_temp['Postal_Code_Encoded'].fillna(train_mean)
    X_test_temp['Postal_Code_Encoded'] = X_test_temp['Postal_Code_Encoded'].fillna(train_mean)
    
    # Concatenate and restore original order
    X_temp = pd.concat([X_train_temp, X_test_temp]).sort_index()
    y_temp = pd.concat([y_train_temp, y_test_temp]).sort_index()
    
    X_temp = X_temp.drop(columns=['Postal_Code'])
    X_final = X_temp

# Add additional features with proper index alignment
if 'condition of the house' in df.columns:
    X_final['condition'] = df['condition of the house'].iloc[X_final.index]

if 'grade of the house' in df.columns:
    X_final['grade'] = df['grade of the house'].iloc[X_final.index]

# Ensure proper data types
if 'Postal_Code_Encoded' in X_final.columns:
    X_final['Postal_Code_Encoded'] = X_final['Postal_Code_Encoded'].astype(float)

# Display information
print(f"Final feature matrix shape: {X_final.shape}")
print(f"Feature dtypes:\n{X_final.dtypes.value_counts()}")

# Handle missing values
missing_vals = X_final.isnull().sum()
if missing_vals.sum() > 0:
    print(f"Missing values found:")
    print(missing_vals[missing_vals > 0])
    for col in X_final.columns:
        if X_final[col].isnull().sum() > 0:
            X_final[col] = X_final[col].fillna(X_final[col].median())

print("\nFinal feature set ready for XGBoost:")
print(X_final.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
import xgboost as xgb

if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    split_date = df['Date'].quantile(0.8)
    
    train_mask = df['Date'] <= split_date
    test_mask = df['Date'] > split_date
    
    X_train = X_final[train_mask]
    X_test = X_final[test_mask]
    y_train = y[train_mask]
    y_test = y[test_mask]
    
    print(f"Time-based split:")
    print(f"Train period: {df[train_mask]['Date'].min().date()} to {df[train_mask]['Date'].max().date()}")
    print(f"Test period: {df[test_mask]['Date'].min().date()} to {df[test_mask]['Date'].max().date()}")

elif 'Neighborhood_Cluster' in X_final.columns:
    X_train, X_test, y_train, y_test = train_test_split(
        X_final, y, 
        test_size=0.2, 
        random_state=42,
        stratify=X_final['Neighborhood_Cluster']
    )
    
    print(f"Stratified split by Neighborhood_Cluster")
    print(f"Train clusters: {X_train['Neighborhood_Cluster'].unique()[:5]}...")
    print(f"Test clusters: {X_test['Neighborhood_Cluster'].unique()[:5]}...")

else:
    X_train, X_test, y_train, y_test = train_test_split(
        X_final, y, 
        test_size=0.2, 
        random_state=42
    )

print(f"\nDataset split:")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")

dtrain = xgb.DMatrix(X_train, label=y_train)
dtest = xgb.DMatrix(X_test, label=y_test)

print(f"\nXGBoost DMatrix created:")
print(f"Training samples: {dtrain.num_row()}")
print(f"Test samples: {dtest.num_row()}")
print(f"Features: {dtrain.num_col()}")
```

```python
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

params = {
    'objective': 'reg:squarederror',
    'eval_metric': 'rmse',
    'learning_rate': 0.05,
    'max_depth': 6,
    'min_child_weight': 5,
    'subsample': 0.8,
    'colsample_bytree': 0.8,
    'gamma': 0.1,
    'alpha': 0.1,
    'lambda': 1,
    'random_state': 42,
    'n_jobs': -1
}

evals = [(dtrain, 'train'), (dtest, 'eval')]
model = xgb.train(
    params=params,
    dtrain=dtrain,
    num_boost_round=500,
    evals=evals,
    early_stopping_rounds=50,
    verbose_eval=50
)
```
